# Data Quality Assessment and Preprocessing

Note: While null datas are often imputed in the preprocessing stage, this may lead to my numerical and graphical in EDA to be flawed. Therefore, I decided to transfer null imputations to the end of my EDA .ipynb. 

```python
# Check for duplicated CustomerID 

print(df['CustomerID'].duplicated().any())
```

```python
unique_vals = {
    col: df[col].unique()
    for col in df.select_dtypes(include=['object', 'category']).columns
}

# print(unique_vals)

# MaritalStatus and HandsetPrice contains 'Unknown' => convert them to nan
df.replace('Unknown', np.nan, inplace=True)
```

```python
print(f"Column(s) where there are only 1 unique values: {[col for col in df if df[col].nunique() == 1]}")

# recheck
print(df['Homeownership'].value_counts())

# The value for Homeownership is always 0 so they are not meaningful 
# Drop the column 'Homeownership' from df

df = df.drop('Homeownership', axis=1)
```

```python
print(f"Number of unique values for ServiceArea: {df['ServiceArea'].nunique()}")

# There are too many ServiceArea so we will convert them to just include cities
df['ServiceCity'] = df['ServiceArea'].str[:3]

print(f"Number of unique values for ServiceCity: {df['ServiceCity'].nunique()}")
```

```python
df['ServiceArea'].unique()
```

```python
# Check if negative values are valid 
for col in df.columns:
    if pd.api.types.is_numeric_dtype(df[col]):
        if (df[col] < 0).any():
            print(col)
```

```python
# Assuming that total recurring charge is negative for refunds. 
# CurrentEquipmentDays being negative does not make logical sense so we will remove.

df['CurrentEquipmentDays'] = df['CurrentEquipmentDays'].clip(lower=0)
```

```python
# Count total rows in the dataframe
print(f"Total rows in dataframe: {len(df)}") 

# Count the percentage of rows where AgeHH1 == 0
print(f"% of rows in dataframe where AgeHH1 is 0: {round((df['AgeHH1'] == 0).sum() / len(df) * 100, 2)}") 

# Count the percentage of rows where AgeHH1 == 0 AND ChildrenInHH == "No"
print(f"% of rows in dataframe where AgeHH1 is 0 and ChildrenInHH is 'No': {round(((df['AgeHH1'] == 0) & (df['ChildrenInHH'] == 'No')).sum() / len(df) * 100, 2)}") 

# Note: There are households where age of head household is 0 but there are children.
# Secondary sources did not explain why AgeHH1 is 0 so we will assume that these are meant to be null. 
df.loc[df['AgeHH1'] == 0, 'AgeHH1'] = np.nan
print("AgeHH1 = 0 finished converting to null")
```

```python
# convert income group to categorical

df['IncomeGroup'] = df['IncomeGroup'].astype('category')
```

```python
# (Umayaparvathi and Iyakutti 1070) mentioned that IncomeGroup = 0 indicates missing data
# So, convert IncomeGroup = 0 to nan

df['IncomeGroup'] = df['IncomeGroup'].copy()
df.loc[df['IncomeGroup'] == 0, 'IncomeGroup'] = np.nan
```

```python
# Convert certain numerical column to categorical

to_be_converted = ['HandsetWebCapable','TruckOwner', 'RVOwner', 'ChildrenInHH']

# Convert 1s to True and 0s to False
for col in to_be_converted:
    df.loc[df[col] == 0, col] = "False"
    df.loc[df[col] == 1, col] = "True"

# Check
for col in to_be_converted:
    print(df[col].value_counts())
```

```python
for col in df:
    df.loc[df[col] == 'Unknown', col] = np.nan    
```

```python
def columns_with_mixed_types(df):
    mixed_cols = []
    for col in df.columns:
        # Get unique types in the column (excluding NaNs)
        types = df[col].dropna().map(type).unique()
        if len(types) > 1:
            mixed_cols.append((col, types))
    return mixed_cols

mixed_columns = columns_with_mixed_types(df)
print("Columns with mixed types and their types:")
for col, types in mixed_columns:
    print(f"{col}: {types}")
```

# Feature Engineering

### Service Quality:

```python
# Feature Engineering
df['TotalCalls'] = df[['ReceivedCalls', 'OutboundCalls', 'InboundCalls']].sum(axis=1)
```

```python
df['CallRatio'] = df['OutboundCalls'] / (df['InboundCalls'] + 1)  # +1 to avoid div by zero
```

```python
df['CallFailureRate'] = (df['DroppedCalls'] + df['BlockedCalls'] + df['UnansweredCalls']) / (df['TotalCalls'] + 1)
```

```python
df['CustomerCareRatio'] = df['CustomerCareCalls'] / (df['TotalCalls'] + 1)
```

```python
df['RetentionCallsPerMonth'] = df['RetentionCalls'] / (df['MonthsInService'] + 1)
```

```python
df['RetentionEffortRatio'] = df['RetentionCalls'] / (df['TotalCalls'] + 1)
```

```python
df['RetentionOfferAcceptanceRate'] = df['RetentionOffersAccepted'] / (df['RetentionCalls'] + 1)
```

```python
print(df['MonthsInService'].max())
print(df['MonthsInService'].min())
df['LoyaltyGroup'] = pd.cut(
    df['MonthsInService'],
    bins=[-1, 6, 12, 24, float('inf')],
    labels=['<6m', '6-12m', '1-2y', '2y+']
)
print(df['MonthsInService'].isnull().sum())
df['LoyaltyGroup'].isnull().sum()
```

```python
df['RetentionSuccessRate'] = df['RetentionOffersAccepted'] / (df['RetentionCalls'] + 1)
```

```python
df['ServiceIssues'] = df[['DroppedCalls', 'BlockedCalls', 'UnansweredCalls', 'DroppedBlockedCalls']].sum(axis=1)
print(df['ServiceIssues'].describe())

# Create bins and labels for ServiceIssueLevel from .describe()
bins = [-1, 0, 30, 100, df['ServiceIssues'].max()]
labels = ['None', 'Low', 'Moderate', 'High']
df['ServiceIssueLevel'] = pd.cut(df['ServiceIssues'], bins=bins, labels=labels)
```

```python
df['ServiceIssueRate'] = df['ServiceIssues'] / (df['MonthsInService'] + 1)
```

### Equipments & Device:

```python
df['ServiceAgeInDays'] = df['MonthsInService'] * 30
df['EquipmentAgeRatio'] = df['CurrentEquipmentDays'] / (df['ServiceAgeInDays'] + 1)
```

```python
df['HasFamily'] = (df['AgeHH1'].notna()) & (df['AgeHH2'].notna())
```

```python
df['CurrentEquipmentDays'].describe()

# Create bins using information from .describe()
df['DeviceAgeGroup'] = pd.cut(
    df['CurrentEquipmentDays'],
    bins=[-1, 180, 365, 730, np.inf],
    labels=['<6 months', '6–12 months', '1–2 years', '2+ years']
)
```

```python
df['HasMultipleDevices'] = (df['Handsets'] > 1).astype(int)
```

```python
df['DeviceToServiceAgeRatio'] = df['CurrentEquipmentDays'] / (df['ServiceAgeInDays'] + 1)
```

```python
df['IsNewCustomer'] = (df['MonthsInService'] < 3).astype(int)
```

### Revenue:

```python
df['RevenuePerMinute'] = df['MonthlyRevenue'] / (df['MonthlyMinutes'] + 1)
```

```python
df['RevenuePerActiveSub'] = df['MonthlyRevenue'] / (df['ActiveSubs'] + 1)
```

```python
# Example: Flag large drop in revenue
df['RevenueDropFlag'] = (df['PercChangeRevenues'] < -50).astype(int)
```
